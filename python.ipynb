## While Loop

```python
#reverse number
a = int(input("Enter number:- "))
rev = 0
while a > 0:
  rev = rev*10 + a%10
  a = a// 10
print(rev)
```

```python
#palindrome using while loop
a = int(input("Enter number:- "))
copy = a
rev = 0
while a > 0:
  rev = rev*10 + a%10
  a = a//10
if rev == copy:
  print("palindrome")
else:
  print("not a palindrome")
```

```python
# Guessing a number
import random
com = random.randint(1,100)
tries = 0
while True:
  tries += 1
  hum = int(input("enter a number: "))
  if com == hum:
    print(f"congratulations you won in {tries} tries !")
    break
  elif com > hum:
    print("choose little larger !")
  elif com < hum:
    print("choose little smaller !")

```

# Functions

Function is a reusable block of code with a name. instead of writing the same logic 10 times, you write it once as a function and call it 10 times


```python

def palindrome(a):
  copy = a
  rev = 0
  while a > 0:
    rev = rev*10 + a%10
    a = a//10
  if rev == copy:
    print("palindrome")
  else:
    print("not palindrome")
palindrome(121)
palindrome(1221)
```

#### how Exception Error solve





1. try --> Put code here that might cause an error. You're "testing" it.

2. except--> If an error happens in the try block, this part runs to handle it.

3. else--> Only runs if no errors happened in the try block.

4. finally--> Always runs, no matter what—even if there was an error or not.

```python
a = int(input("enter 1st number: "))
b = int(input("enter 2nd number: "))

try:
  print(a/b)
except Exception as err:
  print(f"sorry an error occure as {err}")
else:
  print("error not occure")
finally:
  print("always run no matter what ")

age = int(input('enter your age: '))
if age < 18:
  raise ValueError('your not eligible')  #raise error here we form error
print("your are eligible")

name = input('Enter your name: ')
print(f"hello {name} welcome to learn Exception Handl4ing ")
```
